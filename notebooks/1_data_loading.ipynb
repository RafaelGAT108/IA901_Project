# Downloading Lung Sounds Datasets

Lung sounds are recordings captured from the chest with a digital stethoscope or a microphone. They can reveal useful clinical patterns, such as crackles and wheezes, that are not always easy to detect by ear alone, and help to identify various diseases and clinical conditions.

```python
import os
import zipfile
import requests
```

```python
DATA_FOLDER = os.path.abspath("../data/raw")
if not os.path.exists(DATA_FOLDER):
    os.makedirs(DATA_FOLDER)
```

## Datasets

### ICBHI 2017

- An open access database for the evaluation of respiratory sound classification algorithms
 (https://iopscience.iop.org/article/10.1088/1361-6579/ab03ea).
- Data available at: https://bhichallenge.med.auth.gr/ICBHI_2017_Challenge.

#### Data Loading

Downloading the dataset

```python
dataset1_path = os.path.join(DATA_FOLDER, "ICBHI")

if not os.path.exists(dataset1_path):
    os.makedirs(dataset1_path)
    print("Downloading ICBHI dataset...")
    url = "https://bhichallenge.med.auth.gr/sites/default/files/ICBHI_final_database/ICBHI_final_database.zip"
    response = requests.get(url, timeout=30, verify=False)
    response.raise_for_status()

    zip_path = os.path.join(DATA_FOLDER, "ICBHI_final_database.zip")
    with open(zip_path, "wb") as f:
        f.write(response.content)

    print("Extracting ICBHI dataset...")
    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(dataset1_path)
    os.remove(zip_path)
    print("Done!")
else:
    print("ICBHI dataset already exists. Skipping download.")
```

Downloading extra files that contains demographic informations, diagnosis and training/test splits.

```python
extra_files = [
    "ICBHI_Challenge_demographic_information.txt",
    "ICBHI_Challenge_diagnosis.txt",
    "ICBHI_challenge_train_test.txt",
]

for file_name in extra_files:
    file_path = os.path.join(dataset1_path, file_name)
    if os.path.exists(file_path):
        continue

    print(f"Downloading {file_name}...")
    url = f"https://bhichallenge.med.auth.gr/sites/default/files/ICBHI_final_database/{file_name}"
    response = requests.get(url, timeout=30, verify=False)
    response.raise_for_status()
    with open(file_path, "wb") as f:
        f.write(response.content)

print("All required ICBHI files are available.")
```

### KAUH 2021

- A dataset of lung sounds recorded from the chest wall using an electronic stethoscope (https://www.sciencedirect.com/science/article/pii/S2352340921001979).
- Data available at: https://data.mendeley.com/datasets/jwyy9np4gv/3.

#### Data Loading

```python
dataset2_path = os.path.join(DATA_FOLDER, "KAUH")

if not os.path.exists(dataset2_path):
    os.makedirs(dataset2_path)
    print("Downloading KAUH dataset...")
    url = "https://data.mendeley.com/public-api/zip/jwyy9np4gv/download/3"
    response = requests.get(url, timeout=30, verify=False)
    response.raise_for_status()

    zip_path = os.path.join(DATA_FOLDER, "jwyy9np4gv-3.zip")
    with open(zip_path, "wb") as f:
        f.write(response.content)

    print("Extracting KAUH dataset...")
    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(dataset2_path)
    os.remove(zip_path)
    print("Done!")
else:
    print("KAUH dataset already exists. Skipping download.")
```

```python
if not os.path.exists(dataset2_path+"/Audio Files"):
    print("Extracting KAUH audio files...")
    with zipfile.ZipFile(f"{dataset2_path}/Audio Files.zip", "r") as zip_ref:
        zip_ref.extractall(dataset2_path+"/Audio Files")
    os.remove(f"{dataset2_path}/Audio Files.zip")
else:
    print("KAUH audio files already exist. Skipping extraction.")

if not os.path.exists(dataset2_path+"/Stethoscope Files"):
    print("Extracting KAUH stethoscope files...")
    with zipfile.ZipFile(f"{dataset2_path}/Stethoscope Files.zip", "r") as zip_ref:
        zip_ref.extractall(dataset2_path+"/Stethoscope Files")
    os.remove(f"{dataset2_path}/Stethoscope Files.zip")
else:
    print("KAUH stethoscope files already exist. Skipping extraction.")
print("Done!")
```
